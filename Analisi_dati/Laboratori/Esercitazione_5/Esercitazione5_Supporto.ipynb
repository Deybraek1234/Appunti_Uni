```python
#Esempio su un semplice vettore per calcolo Erlang

#genero un semplice array
p = np.arange(9)
print(f"Array di partenza {p}")

#valori di indici pari
pari = p[::2]
print(f"Sottoinsieme dell'array iniziale con indici pari:\t {pari}")


#valori di indici dispari
dispari = p[1::2]
print(f"Sottoinsieme dell'array iniziale con indici dispari:\t {dispari}")


#sommo i dispari ai pari
#nel caso l'array dei pari sia più lungo di quello dei dipari ignoro l'ultimo pari
if(len(pari)>len(dispari)):
    pari = pari[:-1]
    print(f"\nPer rendere la lunghezza di pari e dispari uguale ho tolto l'ultimo dei pari")
    print(f"Il nuovo array pari è {pari}")
risultato = pari+dispari

print(f"\nArray ottenuto sommando gli elementi consecutivi a due a due {risultato}")
```

```python
#Esempio su un semplice vettore per calcolo Poisson

#array di prova, supponiamo che indichino secondi come i tempi di attesa dati dal geiger
p = np.asarray([3,9,5,11])
print(f"Array di prova: {p} secondi")

delta_t = 10
print(f"Voglio determinare i conteggi ogni {delta_t} secondi")

#calcolo la somma cumulativa
tcum = np.cumsum(p)
print(f"Somma cumulativa dell'array di prova: {tcum}")

#determino il minimo e il massimo tempo trascorso
tmin = 0
tmax = tcum[-1]


print(f"totale tempo trascorso {tmax}")

#creo un vettore di finestre a passi di delta_t che userò come tempo di inizio delle mie finestre temporali
w = np.arange(tmin,tmax,delta_t)
print(f"Tempo di inizio delle finestre temporali che userò: {w}")

#controllo se l'ultima finestra temporale è stata conclusa:
if (tmax<w[-1]+delta_t):
    print("Tolgo l'ultima finestra")
    w = w[:-1]

#faccio un loop sulle finestre
Poisson_counts = []

for i in w:
    #verifico se in ogni finestra ho conteggi 
    ind = np.where((tcum>=i) & (tcum<=(i+delta_t)))
    print(f"\nNella finestra ({i},{i+delta_t}) s, i tempi scelti sono {tcum[ind]} quindi un totale {len(tcum[ind])} eventi")
    Poisson_counts.append(len(tcum[ind]))

print(f"\nI conteggi della mia Posson sono {Poisson_counts}")    

```
