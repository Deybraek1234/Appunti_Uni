## Esercizio
#### Lanciamo una moneta 10 volte, quale è la probabilità di ottenere esattamente 6 teste?

## ---
Dato che una moneta può dare testa o croce a ogni lancio, la probabilità che esca testa a ogni lancio è p = 0.5. Useremo la distribuzione binomiale con k=6, n = 10 e p = 0.5

```python
# il primo passaggio è importare i pacchetti necessari 
import math
```

```python
#calcolo la distribuzione di probabilità binomiale usando la formula matematica
k = 6
n = 10
p = 0.5

coeff_binomiale = math.comb(n, k)   # funzione di math che permette di calcolare il coefficiente binomiale
P = coeff_binomiale * (p**k) * ((1 - p)**(n - k))
```

```python
print(f"La probabilità di ottenere {k} volte testa da {n} lanci della moneta è {P}")
```

### Che vuol dire questo risultato?
Vuol dire che se ripeto molte volte 10 lanci di una moneta, nel 20% dei casi ottengo 6 teste

```python
#semplice simulazione di 10 lanci della moneta
import random

for i in range(10):
    lancio = random.randint(0, 1)
    if lancio == 0:
        print("Testa")
    else:
        print("Croce")
```

```python
#aggiungo al codice di prima il conto di quante volte esce testa e quante volte esce croce
testa = 0
croce = 0
for i in range(10):
    lancio = random.randint(0, 1)
    if lancio == 0:
        testa+=1
    else:
        croce+=1
print(f"Ho ottenuto {testa} volte Testa e {croce} volte Croce")        
```

```python
#ora ripeto 100 volte e vedo quante volte ho testa
testa_l = [] #qui salverò quante volte esce testa in ognuno degli esperimenti

for i in range(100): #100 esperimenti
    testa = 0 #all'inizio di ogni esperimento azzero il contatore delle volte che esce testa
    for j in range(10): #Ogni esperimento consiste in 10 lanci della moneta
        lancio = random.randint(0, 1)
        if lancio == 0:
            testa+=1
    #dopo aver finito i 10 lanci salvo il risultato
    testa_l.append(testa)


#ora che ho finito i 100 esperimenti conto quante volte ho ottenuto 6 teste    
print(testa_l.count(6))
```
